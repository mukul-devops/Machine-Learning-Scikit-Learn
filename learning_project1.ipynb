## **Project(Student Marks Predictor)** Based on Supervise Learning with model MAE,MSE,RMSE,R2 SCORE

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

model = LinearRegression()

df = pd.read_csv('student_score_df.csv')

X = df[['Study Hours']]
y = df['Final Score']

model.fit(X,y)

predicted_score = model.predict(X)

mae = mean_absolute_error(predicted_score,y)
mse = mean_squared_error(predicted_score,y)
rmse = np.sqrt(mse)
r2 =r2_score(predicted_score,y)

print("Mean Absolute Error:",round(mae,2))
print("Mean Squared Error:",round(mse,2))
print("Root Mean Squared Error:",round(rmse,2))
print("R^2 Score (Model Accuracy):",round(r2,4))   # closer to 1 is better for model.

print("\nHistogram plot for cheaking distribution of final score")
fig, ax = plt.subplots(figsize=(10,6),dpi=100)
sns.set_theme()
sns.histplot(
            data=df,
            x="Final Score",
            bins=20,
            kde=True,
            alpha=0.6,
            color='c',
            ax=ax
        )

ax.set_title("Distribution of Final Score")
ax.set_xlabel("Exam Score")
ax.set_ylabel("No. of Student")
plt.show()

print("\nScatter Plot+ Regession line(For Predicted score)")
fig, ax = plt.subplots(figsize=(10,6),dpi=100)
ax.scatter(X,y,label='Actual Score')
ax.plot(X,predicted_score,color='r',label='Predicted Score (Regession line)')

ax.set_title('Model Prediction VS Actual Score')
ax.set_xlabel('Study Hours')
ax.set_ylabel('Final Score')
ax.legend()
plt.show()

hour_studied = float(input("Enter how much Hour you study per day:"))
final_predition = model.predict([[hour_studied]])
print(f"Predicted Score Based on your Hour Studed: ({hour_studied}), you may score:{final_predition[0]:.2f}")
```
