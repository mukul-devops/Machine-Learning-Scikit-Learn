## ***Supervised Learning***

**1. Linear Regression**: It is a supervised learning model used to predicts number using straight line relationship between input and output of old data.

- Find a pattern in old data 
- straight line 
- y = mx+c

```python
'''
#Basic Syntax:
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X,y)
model.predict([[value]])   #predict function take 2d list as an input due to model memory storage.

'''
from sklearn.linear_model import LinearRegression

model = LinearRegression()

X = [[1],[2],[3],[4],[5]]
y = [40,50,75,85,90]

model.fit(X,y)

hours = float(input("Enter how many hours you studied: "))

predicted_marks = model.predict([[hours]])
print(f'Based on your study hours{hours} you may score around {predicted_marks}')
```

**2. Classification**

 a. Logistic Rgression: Simple, interpretable, good for binary tasks.

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()

X =[[1],[2],[3],[4],[5]]   # hour studies
y = [0,0,1,1,1]             # Result 0-F, 1-P
model.fit(X,y)
hours  = float(input("Enter how many hr you study:"))

result = model.predict([[hours]])[0]

if result ==1:
    print(f'Based on your hr {hours}, you likely to Pass')
else:
    print(f'Based on your hr {hours}, you likely to Fail')
```

**b. K-NN (K-Nearest Neighbour)**

K-NN classifies a new data point based on the majority class of its nearest neighbors in the training set.

```python
from sklearn.neighbors import KNeighborsClassifier
model = KNeighborsClassifier()

# Data of Weight(gm) and Size(cm) of fruits
X = [
    [100,7],
    [200,7.5],
    [250,8],
    [300,8.5],
    [330,9],
    [360,9.5]
]

# 0 for apple, 1 for orange
y =[0,0,0,1,1,1]

model.fit(X,y)

weight = float(input("Enter Weight(gm) of your fruit:"))
size = float(input("Enter Size(cm) of your fruit:"))

r = model.predict([[weight,size]])

if r==0:
    print(f"May it is Apple")
else:
    print(f"May it is Orange")
```

**c. Decision Tree**

A Decision Tree is a supervised learning algorithm that works like a flowchart, breaking down data into smaller subsets based on feature values until a decision (prediction) is made.



```python
from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier()

X = [
    [7,2],
    [8,3],
    [9,8],
    [10,9]
]

y = [0,0,1,1] #0 for apple and 1 for orange

model.fit(X,y)

size = float(input("Enter the fruit size in cm:"))
shade = float(input("Enter a color shade(0-10):"))

result = model.predict([[size,shade]])[0]

if result==1:
    print("This is likey an apple.")
else:
    print("This is likey an orange.")
```

## ***Model Evaluation & Metrics (Accuracy, Precision, Recall, F1)***

**Accuracy** = actual value /pridicted value

**precision** = measures how accurate a model’s positive predictions are. In other words, when the model says “yes,”
 how often is it correct. It is calculated as the ratio of true positives (TP) to all predicted positives (TP + FP)

- Precision = TP / (TP + FP) .
   - True Positives (TP): Correctly predicted positive cases.
   - False Positives (FP): Incorrectly predicted positive cases (false alarms).
- Precision focuses on reducing false positives.

**Recall** = measures how well a model captures all the actual positive cases
   - Recall = TP / (TP + FN)
        - True Positives (TP): Correctly predicted positive cases.
        - False Negatives (FN): Actual positives that the model missed.
- Recall focuses on reducing false negatives.

**F1 score** = F1-score in machine learning is a metric that combines both precision and
 recall into a single value, giving a balanced measure of a model’s performance.

 - F1 = 2(precision.recall/precision+recall)
- It is the harmonic mean of precision and recall.

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

#True answers (what actual happen)
y_true = [1,0,1,1,0,0]

#Model's Predicted answers(What is guessed)
y_predict = [1,0,1,1,1,1]

print("Accuracy",accuracy_score(y_true,y_predict))
print("Precision",precision_score(y_true,y_predict))
print("Recall",recall_score(y_true,y_predict))
print("F1",f1_score(y_true,y_predict))
```

**Confution Matrix** : [[TP   FN],
                        [FP   TN]]

    TN - True Positive , FP - False Negative
    FN - False Positive, TP - True Negative

```python
from sklearn.metrics import confusion_matrix

y_true = [1,0,1,1,0,0,1,0]
y_pred = [1,0,1,0,0,1,1,0]

cm = confusion_matrix(y_true,y_pred)

print("Confusion Matrix")
print(cm)
```

 **MAE, MSE, RMSE**
 - MAE(Mean Absolute Error): 1. Take mistakes differences.
 2.then take mean of them.

 - MSE(Mean Squared Error): 1. Take mistakes differences.
    2. square them .
    3. Then take mean of them.

 - RMSE(Root Mean Squared Error): take root of MSE.

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

#Real Scores
real_scores = [90,60,80,100]
predicted_scores = [85,70,70,95]

#Model Guess
mae = mean_absolute_error(real_scores,predicted_scores)
mse = mean_squared_error(real_scores,predicted_scores)
rmse = np.sqrt(mse)

print("MAE: ON average off by",mae)
print("MSE: Squared mistake error",mse)
print("RMSE: Final Realistic error",rmse)
```
